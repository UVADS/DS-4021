## Part 3: Histogram-Based Gradient Boosting (20 points)

Gradient Boosting is a powerful algorithm, but it can also be computationally expensive. To address this, a modern and optimized implementation inspired by *LightGBM* has been introduced (see recommended reading). In `scikit-learn`, this approach is available through `HistGradientBoostingClassifier`.

Unlike the traditional `GradientBoostingClassifier`, which searches for the best split exactly, the histogram-based version accelerates training by:
- Binning continuous features into discrete intervals (histograms) to reduce computation.  
- Handling missing values natively, without requiring imputation.  
- Supporting early stopping, allowing the model to automatically stop boosting when no further improvement is achieved.
