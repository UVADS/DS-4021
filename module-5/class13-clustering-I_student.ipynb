## 1. What is Unsupervised Learning? What is Clustering?

Clustering is a common task of what is considered **unsupervised learning**. In contrast to  **supervised learning**, where we have labels (the outcome variable $y$) that "guide" the learning process, in unsupervised learning we only have the input data (the $X$), with no labels at all. The goal here is more to discover hidden structure or patterns within the data.

```python
import numpy as np
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt

X, y = make_blobs(n_samples=300, centers=[[-3, 0], [3, 0]], cluster_std=0.8, random_state=42)

fig, axs = plt.subplots(ncols=2, figsize=(10, 5))
axs[0].scatter(X[y==0][:, 0], X[y==0][:, 1], c="r", marker="+", s=100, linewidths=2)
axs[0].scatter(X[y==1][:, 0], X[y==1][:, 1], c="green", marker="*",  s=100, linewidths=2)
axs[0].set_title("Supervised Learning", size=20)
axs[0].set_xlabel("Feature 1", size=18)
axs[0].set_ylabel("Feature 2", size=18)

axs[1].scatter(X[:, 0], X[:, 1], s=100, linewidths=2, edgecolors="k")
axs[1].set_title("Unsupervised Learning", size=20)
axs[1].set_xlabel("Feature 1", size=18)
axs[1].set_ylabel("Feature 2", size=18)
plt.tight_layout()
```

But not only clustering belongs to unsupervised learning. Dimensionality reduction such as PCA is another type of unsupervised learning task.

Clustering is the process of organizing data into subgroups (called *clusters* or *subtypes*) such that:

- Points within the same cluster are more similar to each other.
- Points from different clusters are less similar.

Since points within the clusters are similar to each other, they are homogeneous groupings of the data. As a result, clustering are suitable techniques when you suspect your data is heterogeneous. This is a typical situation when you work with human populations.

- Grouping customers by purchasing behavior

## 2. Distance and Similarity

We said that, after clustering, observations are more similar to each other within the same cluster, and less similar with respect to points in different clusters. This gives you the intuitive idea that “closeness” or “similarity” is a foundation concept in clustering. But how do we quantify that?

To measure how close two data points are, we use a **distance metric** $D(\vec{x}_1, \vec{x}_2)$.

Here are some examples: 
- **Euclidean**: $D(\vec{x}_1, \vec{x}_2) = \sqrt{\sum_i (x_{1i} - x_{2i})^2}$ 
- **Manhattan**: $D(\vec{x}_1, \vec{x}_2) = \sum_i |x_{1i} - x_{2i}|$ 
- **Cosine distance**: $D(\vec{x}_1, \vec{x}_2) = 1 -\cos(\theta) = 1-\frac{\vec{x}_1 \cdot \vec{x}_2}{||x_1||\,||x_2||}$

All clustering methods rely on a notion of **similarity**, often derived from a distance.

Let's compare these distances for a simple example.

```python
import numpy as np
from scipy.spatial import distance

# Two points
p1 = np.array([1, 2])
p2 = np.array([4, 8])

euclidean = np.linalg.norm(p1 - p2)
manhattan = np.sum(np.abs(p1 - p2))
cosine = distance.cosine(p1, p2)
print(f"Euclidean distance: {euclidean:.2f}")
print(f"Manhattan distance: {manhattan:.2f}")
print(f"Cosine distance: {cosine:.2f}") # Q: Why D is zero?
```

Note that both measure how far apart the points are — but the numerical values (and sometimes the cluster assignments) will depend on which one we use.

## 3. K-means


Now that we covered the idea of *similarity* and *distance*,  let's explore one of the most widely used clustering algorithms: K-Means.

### Intuition

K-Means groups data into $k$ clusters by minimizing the **within-cluster sum of squared distances**.

Algorithm steps:
1. Choose the number of clusters $k$.
2. Randomly initialize $k$ centroids.
3. Assign each point to the nearest centroid (based on distance).
4. Update each centroid as the mean of its assigned points.
5. Repeat until centroids stop moving (convergence).

The goal is to minimize:


$$J = \sum_{i=1}^{k} \sum_{x \in C_i} ||x - \mu_i||^2$$


where $C_i$ is cluster $i$ and $\mu_i$ its centroid.

Let's implement it manually:

```python
from sklearn.datasets import make_classification, make_blobs
import matplotlib.pylab as plt
X, y = make_classification(n_samples=200, 
                           n_features=2, 
                           n_classes=4, 
                           n_informative=2, 
                           n_redundant=0, 
                           n_repeated=0, 
                           n_clusters_per_class=1,
                           random_state=1234,
                           class_sep=2,
                          )

plt.scatter(X[:,0], X[:,1], c=y, cmap='Set2', s=100, linewidths=2, edgecolors="k")
plt.xlabel("Feature 1", size=18)
plt.ylabel("Feature 2", size=18)
```

## 6. Gaussian Mixture Models

There is one problem with k-means: it assigns each point *hardly* to one cluster using the distance to the mean (centroid).

We can soften this by using a gaussian mixture model (GMM), a probabilistic models that assume all the data points are generated from a mixture of a finite number of Gaussian distributions with unknown parameters. 

It is a generalizion of k-means clustering by incorporating information about the covariance structure of the data as well as the centers of the latent Gaussians.

$$
p(\vec{x}) = \sum_{k=1}^{K} \pi_k \, \mathcal{N}(\vec{x} \mid \vec{\mu}_k, \Sigma_k)
$$

where:
- $\pi_k$ are the mixing coefficients (non-negative and sum to 1),
- $\vec{\mu}_k$ is the mean vector of component $k$,
- $\Sigma_k$ is the covariance matrix of component $k$,
- $\mathcal{N}(\vec{x} \mid \vec{\mu}_k, \Sigma_k)$ denotes the multivariate Gaussian distribution:

$$
\mathcal{N}(\vec{x} \mid \vec{\mu}_k, \Sigma_k)
= \frac{1}{(2\pi)^{D/2} |\Sigma_k|^{1/2}}
  \exp\left[-\frac{1}{2} (\vec{x} - \vec{\mu}_k)^\top
  \Sigma_k^{-1} (\vec{x} - \vec{\mu}_k)\right]
$$

The model parameters $(\pi_k, \vec{\mu}_k, \Sigma_k)$ are estimated using an **Expectation–Maximization (EM)** approach:

- E-step: Estimate the posterior probability that each point belongs to each component:
  $$
  \gamma_{ik} = \frac{\pi_k \, \mathcal{N}(\vec{x}_i \mid \vec{\mu}_k, \Sigma_k)}
  {\sum_{j=1}^{K} \pi_j \, \mathcal{N}(\vec{x}_i \mid \vec{\mu}_j, \Sigma_j)}
  $$

- M-step: Update parameters using these probabilities (soft assignments).

The process repeats until convergence of the log-likelihood:

$$
\mathcal{L} = \sum_{i=1}^{N} \log \left( \sum_{k=1}^{K} \pi_k \,
\mathcal{N}(\vec{x}_i \mid \vec{\mu}_k, \Sigma_k) \right)
$$

In `scikit-learn`, this technique is implemented in the `GaussianMixture` class in `mixture`.
